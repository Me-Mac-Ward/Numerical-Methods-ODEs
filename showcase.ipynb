# Example 1: Exponential growth

```python
from functions_for_import import *
from math import exp, sin, cos

# This can be changed to any function desired and will be called in future numerical method implementations
def derivative(x, y):
    return y

def initialiseStates():
    x0 = 0
    y0 = 1
    h = 0.1
    states = [(x0, y0)]
    steps = 10
    return x0, y0, h, states, steps

x0, y0, h, states, steps = initialiseStates()
eulerStates, leapfrogStates, midpointStates, rungeKuttaStates = runAllMethods(initialiseStates, states, steps, h, derivative, x0, y0)
print("Numerical methods solutions: \n")
compareMethods(eulerStates, leapfrogStates, midpointStates, rungeKuttaStates)
print("\n Exact solutions: \n")
print("     x     |       y")
print("-----------+--------------")
for i in range(len(eulerStates)):
    x = x0 + i*h
    y = exp(x)
    print(f"{x:10.4f} | {y:12.6f}")
```

# Example 2: $\frac{dy}{dx}$ dependent on both x and y
