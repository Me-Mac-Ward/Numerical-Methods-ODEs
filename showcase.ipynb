```python
from functions_for_import import *

def initializeStates():
    x0 = 0
    y0 = 0.5
    h = 0.1
    states = [(x0, y0)]
    steps = 10
    return x0, y0, h, states, steps

def runAllMethods(states, steps, h, x0, y0):
    x0, y0, h, states, steps = initializeStates()
    eulerStates = eulersMethod(states, steps, h, x0, y0)
    x0, y0, h, states, steps = initializeStates()
    leapfrogStates = leapfrogMethod(states, steps, h, x0, y0)
    x0, y0, h, states, steps = initializeStates()
    midpointStates = midpointMethod(states, steps, h, x0, y0)
    x0, y0, h, states, steps = initializeStates()
    rungeKuttaStates = rungeKuttaMethod(states, steps, h, x0, y0)
    return eulerStates, leapfrogStates, midpointStates, rungeKuttaStates

def compareMethods(eulerStates, leapfrogStates, midpointStates, rungeKuttaStates):
    print("     x     |  y (Euler)   | y (Leapfrog) | y (Midpoint) |  y (RK4)   ")
    print("-----------+--------------+--------------+--------------+-------------")
    for i in range(len(eulerStates)):
        x = eulerStates[i][0]
        y_euler = eulerStates[i][1]
        y_leapfrog = leapfrogStates[i][1]
        y_midpoint = midpointStates[i][1]
        y_rk4 = rungeKuttaStates[i][1]
        print(f"{x:10.4f} | {y_euler:12.6f} | {y_leapfrog:12.6f} | {y_midpoint:12.6f} | {y_rk4:12.6f}")

eulerStates, leapfrogStates, midpointStates, rungeKuttaStates = runAllMethods(states, steps, h, x0, y0)
compareMethods(eulerStates, leapfrogStates, midpointStates, rungeKuttaStates)
```
